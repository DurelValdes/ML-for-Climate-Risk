```python
import matplotlib.pyplot as plt
import geopandas as gpd
import numpy as np
from matplotlib.colors import ListedColormap, BoundaryNorm

INDICATOR_COLS = {
    "q5":  ["ind_sec_T1_5",  "ind_sec_T2_5",  "ind_sec_T3_5",  "ind_sec_T4_5"],
    "q25": ["ind_sec_T1_25", "ind_sec_T2_25", "ind_sec_T3_25", "ind_sec_T4_25"],
    "off": ["ind_sec_T1_off","ind_sec_T2_off","ind_sec_T3_off","ind_sec_T4_off"]
}

def plot_historical_drought_maps(df, gdf_region, indicator_type,
                                 year_min=2000, year_max=2024):
    """
    4 cartes trimestrielles avec COULEURS DISCRÈTES très contrastées.
    """

    cols = INDICATOR_COLS[indicator_type]

    # ---- 1. Filtre des années ----
    df_hist = datasec[(datasec["ANNEE"] >= year_min) & (datasec["ANNEE"] <= year_max)].copy()

    # ---- 2. Agrégation maille × année ----
    df_yearly = df_hist.groupby(["NUMERO", "ANNEE"])[cols].max().reset_index()

    # ---- 3. Comptage annuel ----
    drought_counts = df_yearly.groupby("NUMERO")[cols].sum()

    # ---- 4. Coordonnées ----
    coords = df_hist.groupby("NUMERO")[["LAMBX_93", "LAMBY_93"]].first()

    drought_map = coords.join(drought_counts)

    # ---- 5. GeoDataFrame final ----
    gdf_pts = gpd.GeoDataFrame(
        drought_map,
        geometry=gpd.points_from_xy(
            drought_map["LAMBX_93"], drought_map["LAMBY_93"]
        ),
        crs=2154
    )

    # ---- 6. Fond simplifié ----
    gdf_region_simple = gdf_region.copy()
    gdf_region_simple["geometry"] = gdf_region_simple.geometry.simplify(200)

    # ---- 7. Palette ULTRA DISTINCTE ----
    colors = [
        "#f2f2f2",  # 0 années = gris très clair
        "#a6cee3",  # 1–3 = bleu clair
        "#1f78b4",  # 4–7 = bleu foncé
        "#fb9a06",  # 8–12 = orange vif
        "#b30000"   # 13–25 = rouge intense
    ]

    bounds = [0, 1, 4, 8, 13, 26]
    cmap = ListedColormap(colors)
    norm = BoundaryNorm(bounds, cmap.N)

    # ---- 8. Figure ----
    fig, axes = plt.subplots(2, 2, figsize=(17, 13))
    axes = axes.flatten()

    titles = ["T1", "T2", "T3", "T4"]

    for i in range(4):
        ax = axes[i]

        gdf_region_simple.plot(
            ax=ax,
            color="white",
            edgecolor="grey",
            linewidth=0.4
        )

        gdf_pts.plot(
            ax=ax,
            column=cols[i],
            cmap=cmap,
            norm=norm,
            markersize=10,
            edgecolor="black",
            linewidth=0.1,
            alpha=1,
            legend=False
        )

        ax.set_title(f"{titles[i]} — {indicator_type.upper()} — {year_min}-{year_max}")
        ax.set_axis_off()

    # ---- 9. Colorbar ----
    cax = fig.add_axes([0.20, 0.93, 0.60, 0.02])
    sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    sm._A = []

    cbar = fig.colorbar(sm, cax=cax, orientation="horizontal", boundaries=bounds)
    cbar.set_label("Nombre d'années de sécheresse (classes discrètes)")

    plt.tight_layout(rect=[0, 0, 1, 0.90])
    plt.show()
```

```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm

INDICATOR_COLS = {
    "q5":  ["ind_sec_T1_5",  "ind_sec_T2_5",  "ind_sec_T3_5",  "ind_sec_T4_5"],
    "q25": ["ind_sec_T1_25", "ind_sec_T2_25", "ind_sec_T3_25", "ind_sec_T4_25"],
    "off": ["ind_sec_T1_off","ind_sec_T2_off","ind_sec_T3_off","ind_sec_T4_off"],
}

def compute_regional_score(df, indicator_type, year_min=2000, year_max=2024):
    """
    Pour chaque région :
    score = moyenne, sur les mailles de la région, du nombre de trimestres secs sur 2000–2024.
    Retourne un DataFrame indexé par code_insee_de_la_region
    avec colonnes: ['region_name', 'score'].
    """

    cols = INDICATOR_COLS[indicator_type]

    # Filtre années
    df_hist = df[(df["ANNEE"] >= year_min) & (df["ANNEE"] <= year_max)].copy()

    # 1) UNE ligne = maille × année (max sur l'année pour chaque trimestre)
    df_year = df_hist.groupby(["NUMERO", "ANNEE"])[cols].max().reset_index()

    # 2) Sécheresse annuelle = somme des 4 trimestres (0 à 4)
    df_year["annual_drought"] = df_year[cols].sum(axis=1)

    # 3) DroughtCount_i = somme sur toutes les années (0 à 100)
    drought_counts = df_year.groupby("NUMERO")["annual_drought"].sum()

    # 4) Rattacher code région + nom de région à chaque maille
    region_info = df_hist.groupby("NUMERO")[["code_insee_de_la_region", "region"]].first()

    df_maille = (
        drought_counts.to_frame("DroughtCount")
        .join(region_info)
    )

    # 5) Agrégation par région (code INSEE) → moyenne sur les mailles
    regional = (
        df_maille
        .groupby("code_insee_de_la_region")
        .agg(
            score=("DroughtCount", "mean"),
            region_name=("region", "first")
        )
    )

    return regional
```

```python
def plot_regional_ranking(regional_df, indicator_type):
    """
    Barplot horizontal des régions, triées par score décroissant.
    regional_df : DataFrame retourné par compute_regional_score
    """

    ranking = regional_df.sort_values("score", ascending=False)

    plt.figure(figsize=(10, 12))
    plt.barh(ranking["region_name"], ranking["score"], color="firebrick")
    plt.gca().invert_yaxis()

    plt.title(f"Nombre moyen de sécheresses (2000–2024) — Indicateur {indicator_type.upper()}")
    plt.xlabel("Sécheresses moyennes par maille (0–100)")
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.patheffects as path_effects

def plot_regional_choropleth(regional_df, gdf_region, indicator_type,
                             code_reg_geom="code_insee"):

    # Récupération des valeurs
    values = regional_df["score"]

    # Construction automatique des classes (quantiles)
    quantiles = values.quantile([0, 0.25, 0.5, 0.75, 1]).values
    bounds = quantiles
    colors = ["#f2f2f2", "#a6cee3", "#1f78b4", "#fb9a06", "#b30000"]

    cmap = ListedColormap(colors)
    norm = BoundaryNorm(bounds, cmap.N)

    # Harmoniser types
    gdf = gdf_region.copy()
    gdf[code_reg_geom] = gdf[code_reg_geom].astype(str)
    regional_df = regional_df.copy()
    regional_df.index = regional_df.index.astype(str)

    # Merge shapefile + scores
    gdf_plot = gdf.merge(
        regional_df,
        left_on=code_reg_geom,
        right_index=True,
        how="left"
    )

    fig, ax = plt.subplots(figsize=(11, 13))

    # Carte chloroplète
    gdf_plot.plot(
        ax=ax,
        column="score",
        cmap=cmap,
        norm=norm,
        edgecolor="black",
        linewidth=0.8,
        legend=False,
    )

    # -------------------------
    # AJOUT DES NOMS DES RÉGIONS
    # -------------------------
    for _, row in gdf_plot.iterrows():
        geom = row["geometry"]
        if geom is None:
            continue

        # Representative point = meilleur placement du label
        x, y = geom.representative_point().x, geom.representative_point().y

        ax.text(
            x, y,
            row["region_name"],
            ha="center",
            va="center",
            fontsize=9,
            fontweight="bold",
            color="black",
            path_effects=[
                path_effects.Stroke(linewidth=2, foreground="white"),
                path_effects.Normal()
            ]
        )

    ax.set_title(f"Sécheresse régionale moyenne (2000–2024) — {indicator_type.upper()}")
    ax.set_axis_off()

    # Barre de couleur
    cax = fig.add_axes([0.22, 0.07, 0.60, 0.02])
    sm = plt.cm.ScalarMappable(norm=norm, cmap=cmap)
    sm._A = []
    plt.colorbar(
        sm,
        cax=cax,
        orientation="horizontal",
        boundaries=bounds,
        label="Sécheresses moyennes par maille (2000–2024)"
    )

    plt.show()
```

```python
def compute_maille_drought_counts(df, indicator_type, year_min=2000, year_max=2024):
    """
    Retourne un DataFrame contenant :
    - NUMERO
    - DroughtCount (nombre total de trimestres secs sur 2000–2024)
    - argile_niveau
    - region
    """

    cols = INDICATOR_COLS[indicator_type]

    # Filtre sur les années
    df_hist = df[(df["ANNEE"] >= year_min) & (df["ANNEE"] <= year_max)].copy()

    # 1) ligne = maille × année, avec sécheresse par trimestre
    df_year = df_hist.groupby(["NUMERO", "ANNEE"])[cols].max().reset_index()

    # 2) sécheresse annuelle (0 à 4)
    df_year["annual_drought"] = df_year[cols].sum(axis=1)

    # 3) sécheresse cumulée par maille (0 à 100)
    drought_counts = df_year.groupby("NUMERO")["annual_drought"].sum()

    # 4) rattache le niveau d'argile et la région
    maille_info = df_hist.groupby("NUMERO")[["argile_niveau", "region"]].first()

    # 5) dataframe final
    df_maille = drought_counts.to_frame("DroughtCount").join(maille_info)

    return df_maille.reset_index()
```

```python
INDICATOR_COLS = {
    "q5":  ["ind_sec_T1_5",  "ind_sec_T2_5",  "ind_sec_T3_5",  "ind_sec_T4_5"],
    "q25": ["ind_sec_T1_25", "ind_sec_T2_25", "ind_sec_T3_25", "ind_sec_T4_25"],
    "off": ["ind_sec_T1_off","ind_sec_T2_off","ind_sec_T3_off","ind_sec_T4_off"],
}

def compute_regional_drought(df, indicator_type, year_min=2000, year_max=2024):
    """
    Retourne un DF indexé par code_insee_de_la_region
    avec colonnes: region_name, score (moyenne sécheresse par maille).
    """
    cols = INDICATOR_COLS[indicator_type]

    df_hist = df[(df["ANNEE"] >= year_min) & (df["ANNEE"] <= year_max)].copy()

    # 1️⃣ Maille × année
    df_year = df_hist.groupby(["NUMERO", "ANNEE"])[cols].max().reset_index()
    df_year["annual_drought"] = df_year[cols].sum(axis=1)

    # 2️⃣ Sécheresse cumulée par maille
    drought_counts = df_year.groupby("NUMERO")["annual_drought"].sum()

    # 3️⃣ Rattacher région
    region_info = df_hist.groupby("NUMERO")[["code_insee_de_la_region", "region"]].first()

    df_maille = drought_counts.to_frame("DroughtCount").join(region_info)

    # 4️⃣ Score moyen régional
    regional = (
        df_maille
        .groupby("code_insee_de_la_region")
        .agg(
            score=("DroughtCount", "mean"),
            region_name=("region", "first")
        )
    )

    return regional
```

```python
def compute_regional_climate(df):
    """
    Calcule :
    - PRENEI_ann = somme pluie sur 12 mois
    - ETP_ann = somme ETP sur 12 mois
    - ClimatRaw = PRENEI_ann - ETP_ann
    - Standardisation : (ClimatRaw - mu) / sigma
    - Agrégation régionale
    """

    df = df.copy()

    # --- Colonnes mensuelles ---
    pre_cols = [f"PRENEI_{m:02d}" for m in range(1, 13)]
    etp_cols = [f"ETP_{m:02d}" for m in range(1, 13)]

    # --- Indicateurs annuels ---
    df["PRENEI_ann"] = df[pre_cols].sum(axis=1)
    df["ETP_ann"]    = df[etp_cols].sum(axis=1)

    # --- Score brut ---
    df["ClimatRaw"] = df["PRENEI_ann"] - df["ETP_ann"]

    # --- Standardisation nationale ---
    mu = df["ClimatRaw"].mean()
    sigma = df["ClimatRaw"].std()

    df["ClimatScore_std"] = (df["ClimatRaw"] - mu) / sigma

    # --- Agrégation par région ---
    reg = (
        df.groupby(["code_insee_de_la_region", "region"])
          .agg(
              ClimatScore_std=("ClimatScore_std", "mean"),
              ArgileScore=("argile_niveau", "mean")
          )
    )

    # Mise en forme
    reg = reg.reset_index().set_index("code_insee_de_la_region")
    reg = reg.rename(columns={"region": "region_name"})

    return reg
```

```python
def build_final_table(df):
    
    # ClimatScore + ArgileScore
    climate = compute_regional_climate(df)

    # Scores sécheresse (OFF, Q5, Q25)
    off  = compute_regional_drought(df, "off").rename(columns={"score": "OFF_moy"})
    q5   = compute_regional_drought(df, "q5").rename(columns={"score": "Q5_moy"})
    q25  = compute_regional_drought(df, "q25").rename(columns={"score": "Q25_moy"})

    # Fusion
    final = (
        climate
        .join(off["OFF_moy"])
        .join(q5["Q5_moy"])
        .join(q25["Q25_moy"])
    )

    return final
```

Les résultats obtenus montrent que la variabilité spatiale de la sécheresse durable ne peut pas s’expliquer uniquement par les conditions climatiques moyennes. 

À l’échelle régionale, les régions du Sud (Occitanie, Provence‑Alpes‑Côte d’Azur, Nouvelle‑Aquitaine) présentent à la fois un ClimatScore_std plus sec que la moyenne nationale et des valeurs élevées de OFF, Q25 et Q5, ce qui traduit des épisodes de sécheresse à la fois fréquents, durables et parfois extrêmes. 

À l’inverse, des régions comme la Bretagne, le Centre‑Val de Loire ou les Pays de la Loire combinent un climat plus proche de la normale ou légèrement humide, un ArgileScore plus faible et des niveaux de sécheresse nettement plus modérés, ce qui en fait des zones globalement moins exposées. 

Certaines régions du quart Est (Grand Est, Auvergne‑Rhône‑Alpes, Bourgogne‑Franche‑Comté) occupent une position intermédiaire intéressante : malgré un ClimatScore_std plutôt humide, elles présentent des valeurs de OFF et de quantiles élevées, suggérant que la sévérité de la sécheresse y est davantage liée à la structure des sols, à la saisonnalité et à la fréquence des épisodes défavorables qu’au seul bilan annuel Pluie − ETP.

L’analyse conjointe de ClimatScore_std, ArgileScore, OFF, Q25 et Q5 met ainsi en évidence un mécanisme combiné où climat, nature des sols et distribution des épisodes secs contribuent ensemble à façonner la vulnérabilité régionale à la sécheresse.

## Représentation  du climate score standardisé en France

```python
def compute_climate_score_by_maille(df):
    # Colonnes mensuelles pluie et ETP
    pluie_cols = [f"PRENEI_{str(m).zfill(2)}" for m in range(1, 13)]
    etp_cols   = [f"ETP_{str(m).zfill(2)}"    for m in range(1, 13)]

    # Calcul pluie annuelle - ETP annuelle
    df["Pluie_ann"] = df[pluie_cols].sum(axis=1)
    df["ETP_ann"]   = df[etp_cols].sum(axis=1)
    df["ClimatScore"] = df["Pluie_ann"] - df["ETP_ann"]

    # Agrégation par maille
    maille = df.groupby("NUMERO").agg({
        "ClimatScore": "mean",
        "LAMBX_93": "first",
        "LAMBY_93": "first"
    })

    # Standardisation
    maille["ClimatScore_std"] = (
        (maille["ClimatScore"] - maille["ClimatScore"].mean()) /
        maille["ClimatScore"].std()
    )

    return maille
```

### Reponse à la question de l'évolution de la tendance( Grand Est)

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import numpy as np
from scipy import stats

# Définition des colonnes par indicateur
COLS = {
    "OFF": ["ind_sec_T1_off", "ind_sec_T2_off", "ind_sec_T3_off", "ind_sec_T4_off"],
    "Q5":  ["ind_sec_T1_5",   "ind_sec_T2_5",   "ind_sec_T3_5",   "ind_sec_T4_5"],
    "Q25": ["ind_sec_T1_25",  "ind_sec_T2_25",  "ind_sec_T3_25",  "ind_sec_T4_25"]
}

def compute_yearly_series(df, indicator_cols):
    """Renvoie la série annuelle moyenne d'un indicateur sur une région."""
    
    # 1) Max par trimestre sur une même année pour chaque maille
    df_year = (
        df.groupby(["NUMERO", "ANNEE"])[indicator_cols]
        .max()
        .sum(axis=1)   # somme T1+T2+T3+T4
        .reset_index(name="annual_drought")
    )

    # 2) Moyenne régionale annuelle
    return (
        df_year.groupby("ANNEE")["annual_drought"]
        .mean()
        .reset_index()
    )


def test_significance(x, y):
    """Renvoie slope, p_value (test statistique de la pente)."""
    
    x_centered = x - np.mean(x)  # meilleure estimation numérique
    slope, intercept, r, p_value, std_err = stats.linregress(x_centered, y)
    
    return slope, p_value


def plot_three_indicators(df, region_code):

    # --- Filtrer la région ---
    df_reg = df[df["code_insee_de_la_region"] == region_code].copy()
    region_name = df_reg["region"].iloc[0]

    # --- Construire les séries temporelles ---
    ts = {name: compute_yearly_series(df_reg, cols) for name, cols in COLS.items()}

    # --- Préparation du graphique ---
    plt.figure(figsize=(14, 7))
    ax = plt.gca()

    colors = {"OFF": "red", "Q5": "orange", "Q25": "blue"}

    slopes = {}
    significances = {}

    # --- Tracer chaque indicateur ---
    for name in ["OFF", "Q5", "Q25"]:
        df_ts = ts[name]
        x = df_ts["ANNEE"].values
        y = df_ts["annual_drought"].values

        # Courbe seule (tu ne veux pas les droites)
        ax.plot(x, y, marker="o", label=f"{name}", color=colors[name])

        # Test statistique
        slope, p_value = test_significance(x, y)
        slopes[name] = slope
        significances[name] = "✔" if p_value < 0.05 else "✘"

    # --- Mise en forme ---
    plt.title(
        f"Évolution de la sécheresse (2000–2024) — Région {region_name}\n"
        f"Pentes : OFF={slopes['OFF']:.3f}{significances['OFF']} | "
        f"Q5={slopes['Q5']:.3f}{significances['Q5']} | "
        f"Q25={slopes['Q25']:.3f}{significances['Q25']}",
        fontsize=15
    )

    plt.xlabel("Année")
    plt.ylabel("Sécheresse moyenne (nombre de trimestres secs)")
    plt.legend()
    plt.grid(True, linestyle="--", alpha=0.4)
    plt.show()


# --- APPEL ---
plot_three_indicators(datasec, region_code=44)  # Grand Est
```

Sur la période 2000–2024, la région Grand Est présente une forte variabilité interannuelle de la sécheresse.
Certaines années se démarquent nettement, avec des pics prononcés en 2003, 2011, 2018, 2020 et 2022, traduisant des épisodes particulièrement secs.
Ces années contrastent avec plusieurs années très humides (notamment 2000, 2001, 2008, 2013 ou 2023), ce qui renforce le caractère irrégulier du phénomène.

Les tendances estimées pour les indicateurs OFF, Q5 et Q25 sont légèrement positives, ce qui pourrait suggérer une augmentation progressive de la sécheresse.
Cependant, ces pentes ne sont pas statistiquement significatives au seuil de 5%, ce qui signifie qu’aucune aggravation durable ne peut être démontrée.

En résumé, le Grand Est connaît bien des épisodes de sécheresse marqués, concentrés sur quelques années récentes, mais il n’existe pas de tendance robuste indiquant une augmentation structurelle de la sécheresse sur l'ensemble de la période. La dynamique reste dominée par des fluctuations d’une année à l’autre.

### Analyse globale de l'évolution de la sécheresse sur toutes les régions

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# -------------------------------------------------------------------
# 1. Paramètres des indicateurs
# -------------------------------------------------------------------
COLS = {
    "OFF": ["ind_sec_T1_off", "ind_sec_T2_off", "ind_sec_T3_off", "ind_sec_T4_off"],
    "Q5":  ["ind_sec_T1_5",   "ind_sec_T2_5",   "ind_sec_T3_5",   "ind_sec_T4_5"],
    "Q25": ["ind_sec_T1_25",  "ind_sec_T2_25",  "ind_sec_T3_25",  "ind_sec_T4_25"]
}

colors = {"OFF": "red", "Q5": "orange", "Q25": "blue"}


# -------------------------------------------------------------------
# 2. Fonctions utilitaires
# -------------------------------------------------------------------
def compute_yearly_series(df_region, indicator_cols):
    """
    Renvoie la série annuelle moyenne d'un indicateur sur une région.
    - df_region : sous-ensemble d'une seule région
    """
    # Max par trimestre pour chaque maille & année
    df_year = (
        df_region.groupby(["NUMERO", "ANNEE"])[indicator_cols]
        .max()
        .sum(axis=1)                    # somme T1+T2+T3+T4 = nb de trimestres secs
        .reset_index(name="annual_drought")
    )

    # Moyenne régionale annuelle
    return (
        df_year.groupby("ANNEE")["annual_drought"]
        .mean()
        .reset_index()
    )


def test_significance(x, y):
    """
    Pente + p-value (test t classique de la régression linéaire).
    """
    x_centered = x - np.mean(x)
    slope, intercept, r, p_value, std_err = stats.linregress(x_centered, y)
    return slope, p_value


# -------------------------------------------------------------------
# 3. Fonction principale : tracer toutes les régions
# -------------------------------------------------------------------
def plot_all_regions_drought(df):
    """
    Produit une figure avec 1 subplot par région (13 régions métropolitaines),
    alignés dans une grille 4x4. Pour chaque région : OFF, Q5, Q25 + pentes
    et symbole de significativité (✔ / ✘).
    """
    # Liste des régions (code + nom), triées par code
    regions = (
        df[["code_insee_de_la_region", "region"]]
        .drop_duplicates()
        .sort_values("code_insee_de_la_region")
        .reset_index(drop=True)
    )

    n_regions = len(regions)
    ncols = 4
    nrows = int(np.ceil(n_regions / ncols))

    fig, axes = plt.subplots(
        nrows, ncols,
        figsize=(5 * ncols, 4 * nrows),
        sharex=True, sharey=True
    )
    axes = axes.ravel()

    # Boucle sur les régions
    for i, row in regions.iterrows():
        ax = axes[i]
        code_reg = row["code_insee_de_la_region"]
        name_reg = row["region"]

        df_reg = df[df["code_insee_de_la_region"] == code_reg].copy()

        # Construire les séries temporelles pour OFF / Q5 / Q25
        ts = {name: compute_yearly_series(df_reg, cols)
              for name, cols in COLS.items()}

        slopes = {}
        sig = {}

        # Tracer les 3 courbes
        for name in ["OFF", "Q5", "Q25"]:
            df_ts = ts[name]
            x = df_ts["ANNEE"].values
            y = df_ts["annual_drought"].values

            ax.plot(x, y, marker="o", color=colors[name], label=name, linewidth=1.5)

            slope, p_value = test_significance(x, y)
            slopes[name] = slope
            sig[name] = "✔" if p_value < 0.05 else "✘"

        # Titre du subplot
        ax.set_title(
            f"{name_reg}\n"
            f"OFF={slopes['OFF']:.3f}{sig['OFF']} | "
            f"Q5={slopes['Q5']:.3f}{sig['Q5']} | "
            f"Q25={slopes['Q25']:.3f}{sig['Q25']}",
            fontsize=9
        )

        # Grille légère
        ax.grid(True, linestyle="--", alpha=0.3)

        # Axe x seulement sur la dernière ligne
        if i // ncols == nrows - 1:
            ax.set_xlabel("Année")
        if i % ncols == 0:
            ax.set_ylabel("Nb moyen de trimestres secs")

    # Supprimer les axes vides (si 16 subplots pour 13 régions)
    for j in range(n_regions, nrows * ncols):
        fig.delaxes(axes[j])

    # Légende commune
    handles = [plt.Line2D([0], [0], color=colors[n], marker='o', linestyle='-',
                          label=n) for n in ["OFF", "Q5", "Q25"]]
    fig.legend(
        handles=handles,
        loc="lower center",
        ncol=3,
        bbox_to_anchor=(0.5, 0.02)
    )

    fig.suptitle(
        "Évolution de la sécheresse (2000–2024)\n"
        "OFF, Q5, Q25 par région — pentes + significativité (✔ : p<0,05)",
        fontsize=14,
        y=0.98
    )

    plt.tight_layout(rect=[0.02, 0.06, 0.98, 0.94])
    plt.show()


# -------------------------------------------------------------------
# 4. Appel général
# -------------------------------------------------------------------
plot_all_regions_drought(datasec)
```

Sur 2000–2024, les régions françaises montrent une forte variabilité d’une année à l’autre, avec des pics récurrents (2003, 2011, 2018, 2020, 2022) mais aussi plusieurs années très humides.

Les pentes des indicateurs OFF, Q5 et Q25 sont généralement faibles et non significatives, ce qui signifie qu’aucune tendance claire d’aggravation de la sécheresse ne se dégage au niveau national.

Quelques régions du Sud et du Sud-Est présentent des signaux plus marqués, mais sans preuve statistique robuste.

La sécheresse existe, parfois intense, mais reste irrégulière et sans tendance structurelle forte sur l’ensemble du territoire.
